## Windowing
Each sequence needs to to transformed into a set of windows.

For a given encoded sequence *S*, for each windows size *w* in some set of window sizes *W*, return a list (probably a slice-type) of all of windows of size *w* covering *S*.

Use the same step-size for all window sizes. The step-size is the distance (in **encoded nucleotides**)

```python
import sys
import pickle
import gabi.utils as gu
```

```python
datafile = 'data/020_encoded_test_data.p'
samples = pickle.load(open(datafile, "rb"))
sample = samples[3]
print(gu.sample_summary(sample))

print(samples[3]['encoded']['gcmm'].keys())
print(sample['encoded']['gcmm']['seq'].shape)

```

```python
import numpy as np

def window_idx(length:int, w:int, step:int=1) -> np.ndarray:
    # Initialise window at start=0
    start = 0
    end = w
    # step window while within length
    while end <= length:
        # yield current window position
        yield (start, end)
        # update window position
        start = start + step
        end = end + step
    
# test known sequence    
t = np.array(range(0,64))
print(t)
bits = 4
for a,b in window_idx(t.shape[0], bits*5, step=bits):
    c = int((a + b)/2)
    d = int(c / bits)
    print(f"{a}:{b}", f"slice-centre:{c}", f"seq-centre:{d}", t[a:b])


# test sample
sample_seq = sample['encoded']['categorical']['seq']
sample_bits = sample['encoded']['categorical']['bits']
window_size = sample_bits * 100 # 99 nucleotides in window

print("-----------\n\n")
print(f"bits: {sample_bits}")
maxiter = 10
count = 0
for a,b in window_idx(sample_seq.shape[0], window_size, step=sample_bits):
    count += 1
    if count > maxiter: break
    c = int((a + b)/2)
    #d = int(c / sample_bits)
    d = int( (a + b) / (2*sample_bits) )
    print(f"{a}:{b}", f"slice-centre:{c}", f"seq-centre:{d}")#, sample_seq[a:b])
```
